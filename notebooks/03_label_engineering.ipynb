# Label Engineering

The next part of the preparation is to execute label engineering. That is, to extract the heart rate from ECG readings, and calculate subject-wise thresholds in order to re-classify the null class. 

### Table of Contents
0. Import Libraries & Data

The following functions have been implemented in helper_functions.py:

- calculate_tilt: Calculates tilt angle in degrees for each data point according to 3-dimensional filtered data. 
- extract_ar_coeffs: Calculates autoregression coefficients (which indicate how strongly past values of a variable influence its current value).
- consolidate_bpm: Categorizes "valid" BPM values for both L1 and L2 nodes, then calculates average (consolidated) BPM.
- extract_hr_features: Extracts HR from ECG measurements subject by subject to prevent signal bleeding between subjects.
- plot_ecg_signal: Plots ECG signal using mpl.pyplot. 
- find_r_peaks: Finds the r-peaks of the ECG signals using scipy.find_peaks and customized threshold values.
- calculate_sma: Calculates the signal magnitude area for ECG signals. 
- calculate_subject_baselines: Calculates subjects' baseline BPMs according to data from rest activity.
